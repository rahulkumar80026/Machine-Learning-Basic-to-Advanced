Numpy array vs Python lists

```python
# speed
# list
a = [i for i in range(10000000)]
b = [i for i in range(1000000,20000000)]

c = []
import time

start = time.time()
for i in range(len(a)):
    c.append(a[i]+ b[i])
    
print(time.time()-start)
  
```

```python
# numpy
import numpy as np
a = np.arange(10000000)
b = np.arange(10000000,20000000)

start = time.time()
c= a+b
print(time.time()-start)
```

```python
# memory
a = [i for i in range(10000000)]
import sys

sys.getsizeof(a)
```

```python
a = np.arange(10000000,dtype=np.int32)
sys.getsizeof(a)
```

Advance Indexing

```python
# Normal Indexing and slicing
a = np.arange(24).reshape(6,4)
a
```

```python
a[1,2]
```

```python
a[1:3,1:3]
```

```python
# Fancy Indexing
a[[0,2,3]]
```

```python
a[[0,2,3,5]]
```

```python
a[:,[0,2,3]]
```
