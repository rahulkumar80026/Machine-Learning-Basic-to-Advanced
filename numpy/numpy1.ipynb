Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# min/max/sum/prod
# 0 -> col and 1-> row
np.max(a1)
np.sum(a1)
np.prod(a1)
np.max(a1,axis =0)
```

```python
# mean/median/std/var
np.mean(a1, axis = 1)
np.median(a1)
np.std(a1)
np.var(a1)
```

```python
# trignometric function
np.sin(a1)
```

```python
# dot product
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(4,3)
np.dot(a1,a2)
```

```python
# log and exponents
np.log(a1)
np.exp(a1)
```

```python
# round/floor/ceil

np.ceil(np.random.random((2,3))*100)

```

Stacking

```python
# horizontal stacking
a4=np.arange(12).reshape(3, 4)
a5=np.arange(12,24).reshape(3, 4)
np.hstack((a4, a5))
```
