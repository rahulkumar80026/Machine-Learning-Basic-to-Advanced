What is Pandas


Pandas is a fast,powerful,flexible and easy to use open source data analysis and manipulation tool,build on top of the Python programming language


Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

Importing Pandas

```python
import numpy as np 
import pandas as pd
```

Series from lists

```python
# string
country = ['India','Pakistan','USA','Nepal','Srilanka']

pd.Series(country)
```

```python
# integer
runs = [13,24,5,67,78,100]
pd.Series(runs)
```

```python
# custom index
marks =[10,5,3,66,68,100]
subject=['maths','science','oral','physic','english','hindi']

pd.Series(marks,index=subject)
```

```python
# setting a name
marks=pd.Series(marks,index=subject,name='Rahul ke marks')
marks
```

Series from dict

```python
marks = {
    'math':67,
    'english':65,
    'science':52,
    'hindi':95
}

marks_series = pd.Series(marks,name='Rahul ke marks')
marks_series
```
