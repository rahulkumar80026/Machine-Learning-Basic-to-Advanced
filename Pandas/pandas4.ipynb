```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv('imdb-top-1000.csv')
```

```python
movies.head()
```

```python
genres = movies.groupby('Genre')
genres
```

```python
# Applying builtin aggregation function on groupby objects
genres.sum()
```

```python
genres.min()
```

```python
genres.mean(numeric_only=True)
```

```python
# find the top 3 genres by total earning
movies.groupby('Genre').sum()['Gross']
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False)
movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

```python
# second option
movies.groupby('Genre')['Gross']
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```

```python
# find the genre with highest avg IMDB rating
movies.groupby('Genre')['IMDB_Rating'].mean()
movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# find the director with most popular(according to no. of votes)
movies.groupby('Director')['No_of_Votes'].sum()
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated   genre (according to IMDB rating)
movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
# find number of movies done by each actor
# movies['Star1'].value_counts()

movies.groupby('Star1')['Series_Title'].count()
movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```

```python
# GroupBy Attributes and Methods
# find total number of group -> len
# find items in ach group -> size
# first()/last() -> nth item
# get_group -> vs filtering
# groups
# describe 
# sample
# nunique
```

```python
# find how many groups form  in Genre
movies.groupby('Genre')
len(movies.groupby('Genre'))
```

```python
movies['Genre'].nunique()
```

```python
movies.groupby('Genre').size()
```

```python
movies['Genre'].value_counts()
```

```python
# find the every fist movie of Genre
genres = movies.groupby('Genre')
genres.first()
genres.last()

genres.nth(6)
```

```python
# get group
genres.get_group('Horror')
```

```python
genres.get_group('Fantasy')
```

```python
# 2nd option
movies[movies['Genre'] == 'Fantasy']
```

```python
# groups
genres.groups
```

```python
# describe
genres.describe()
```

```python
genres.sample() # it goes in each genre and pick a sample 
genres.sample(2,replace=True)
```

```python
# nunique
genres.nunique()
```

```python
# agg method
# passing dict

# genres.sum() # it will not give meaning insight
genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes':'sum',
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# passing list
genres.agg(['min','max'])
genres.agg(['min','max','sum'])
```

```python
num_cols = movies.select_dtypes(include='number').columns

movies.groupby('Genre')[num_cols].agg(['min','max','mean','sum'])
```

```python
movies.groupby('Genre')['IMDB_Rating'].agg(['min','max','mean','sum'])
```

```python
# Adding both the syntax
genres.agg(
    {
        'Runtime':['mean','min'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross':'sum',
        'Metascore':'min'
    }
)
```
