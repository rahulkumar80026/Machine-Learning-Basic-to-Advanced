3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
y
```

```python
xx , yy = np.meshgrid(x,y)  # it create a grid of x and y values. It is used to create a 2D grid of points from the 1D arrays x and y. The resulting xx and yy arrays will have the same shape, where each element in xx corresponds to a value from x and each element in yy corresponds to a value from y. This is often used for plotting functions or surfaces in 3D space.
xx
```

```python
z = xx**2 + yy**2
z.shape
```

```python

fig = plt.figure(figsize=(12, 8))  # <-- Correct way
ax = fig.add_subplot( projection='3d')
p = ax.plot_surface(xx, yy, z, cmap='viridis')

plt.colorbar(p)
plt.show()
```

```python
z = np.sin(xx) + np.cos(yy)


fig = plt.figure(figsize=(12, 8))  # <-- Correct way
ax = fig.add_subplot( projection='3d')
p = ax.plot_surface(xx, yy, z, cmap='viridis')

plt.colorbar(p)
plt.show()
```

Contour Plots

```python
# Contour Plots (it convert the 3D surface into 2D plot with contour lines)
z = xx**2 + yy**2
fig = plt.figure(figsize=(12, 8))  # <-- Correct way
ax = fig.add_subplot()
contour = ax.contourf(xx, yy, z, cmap='viridis')

fig.colorbar(contour)
```
