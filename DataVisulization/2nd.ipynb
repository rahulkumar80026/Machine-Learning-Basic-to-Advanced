```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Colored Scatterplots

```python
iris = pd.read_csv('iris.csv')
iris.sample(5)
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.8)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

Plot Size

```python
plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.8)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
```

Annotations

```python
batters = pd.read_csv('batter.csv')
```

```python
batters.head()
```

```python
sample_df = batters.head(100).sample(25,random_state = 5)
```

```python
sample_df
```

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
x =[1,2,3,4]
y =[5,6,7,8]

plt.scatter(x,y)
plt.text(1,5,'point 1')
plt.text(2,6,'point 2')
plt.text(3,7,'point 3',fontdict={'size':12,'color':'brown'})
plt.text(4,8,'point 4')
```

Horizontal and verticals lines

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130,color='red')
plt.axvline(30,color='green')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

Subplots

```python
# A diff way to plot graph
batters.head()
```

```python
plt.figure(figsize=(18,10))

plt.scatter(batters['avg'],batters['strike_rate'])
plt.title('Batter Analysis')
plt.xlabel('Average')       
plt.ylabel('Strike Rate')

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(18,10))

ax.scatter(batters['avg'],batters['strike_rate'],color='red',marker='*')
ax.set_title('Batter Analysis')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')

fig.show()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1,sharex=True, figsize=(10,7))

axes[0].scatter(batters['avg'],batters['strike_rate'],color='red')
axes[1].scatter(batters['avg'],batters['runs'])

axes[0].set_title('Avg vs Strike Rate')
# axes[0].set_xlabel('Average')
axes[0].set_ylabel('Strike Rate')
axes[1].set_title('Avg vs Runs')

axes[1].set_title('Avg vs Runs')
axes[1].set_xlabel('Average')
axes[1].set_ylabel('Runs')
```
