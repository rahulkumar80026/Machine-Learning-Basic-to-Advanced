## Question 2: What is the difference between a list and a tuple?

```python
# LIST - Mutable (can be modified)
my_list = [1, 2, 3, 4, 5]
my_list[0] = 10  # Can modify
my_list.append(6)  # Can add elements
print("List:", my_list)

# TUPLE - Immutable (cannot be modified)
my_tuple = (1, 2, 3, 4, 5)
# my_tuple[0] = 10  # This will raise an error
print("Tuple:", my_tuple)

# When to use:
# Use LIST when you need to modify data
# Use TUPLE when data should remain constant
```

## Question 3: How do you create a dictionary and access its values?

```python
# Create a dictionary
person = {
    'name': 'John',
    'age': 30,
    'city': 'New York',
    'email': 'john@example.com'
}

# Access values
print("Name:", person['name'])
print("Age:", person['age'])

# Add new key-value pair
person['phone'] = '123-456-7890'
print("\nUpdated dictionary:")
print(person)

# Other methods
print("\nKeys:", person.keys())
print("Values:", person.values())
```

## Question 4: What is the difference between == and is operators?

```python
# == checks for value equality
# is checks for identity (same object in memory)

a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)  # True (same values)
print("a is b:", a is b)  # False (different objects)
print("a is c:", a is c)  # True (same object)

# For None, always use 'is'
x = None
print("\nx is None:", x is None)  # Correct way
print("x == None:", x == None)   # Works but not recommended
```

## Question 5: What are list comprehensions?

```python
# List comprehension is a concise way to create lists

# Traditional way
squares_traditional = []
for i in range(1, 11):
    squares_traditional.append(i ** 2)
print("Traditional:", squares_traditional)

# List comprehension way
squares = [i ** 2 for i in range(1, 11)]
print("List comprehension:", squares)

# With condition
even_squares = [i ** 2 for i in range(1, 11) if i % 2 == 0]
print("Even squares:", even_squares)
```

## Question 6: What is a function and difference between parameters and arguments?

```python
# Parameters are variables in function definition
# Arguments are actual values passed when calling function

def greet(name, age=25):  # name and age are PARAMETERS (age has default value)
    print(f"Hello {name}, you are {age} years old")

# Calling with arguments
greet("Alice", 30)  # "Alice" and 30 are ARGUMENTS
greet("Bob")  # Uses default age value

# Function with return value
def add(a, b):
    return a + b

result = add(5, 3)
print(f"\nAddition result: {result}")
```

## Question 7: Difference between append() and extend() methods

```python
# append() - adds entire object as single element
list1 = [1, 2, 3]
list1.append([4, 5])
print("After append([4, 5]):", list1)

# extend() - adds each element from iterable
list2 = [1, 2, 3]
list2.extend([4, 5])
print("After extend([4, 5]):", list2)

# Another example
list3 = [1, 2, 3]
list3.extend([4, 5, 6])
print("After extend([4, 5, 6]):", list3)
```

## Question 8: How do you handle exceptions in Python?

```python
# Try-except block for error handling

# Example 1: Handle division by zero
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except Exception as e:
    print(f"An error occurred: {e}")
finally:
    print("This always executes")

# Example 2: Handle type error
try:
    num = int("abc")
except ValueError:
    print("\nError: Cannot convert 'abc' to integer")

# Example 3: Handle index error
try:
    my_list = [1, 2, 3]
    print(my_list[10])
except IndexError:
    print("\nError: Index out of range")
```

## Question 9: What is for loop and while loop?

```python
# FOR LOOP - Used to iterate over sequences (list, tuple, string, etc.)
print("FOR LOOP - Print numbers 1-5:")
for i in range(1, 6):
    print(i, end=" ")

print("\n\nFOR LOOP - Iterate over list:")
fruits = ['apple', 'banana', 'orange']
for fruit in fruits:
    print(f"I like {fruit}")

# WHILE LOOP - Runs as long as condition is true
print("\n\nWHILE LOOP - Print numbers 1-5:")
count = 1
while count <= 5:
    print(count, end=" ")
    count += 1

# When to use:
# FOR LOOP: When you know number of iterations or have a sequence
# WHILE LOOP: When number of iterations depends on a condition
```

## Question 10: What is a module and how do you import modules?

```python
# A module is a file containing Python code (functions, classes, variables)
# You can import modules to reuse code

# Method 1: Import entire module
import math
print("Square root of 16:", math.sqrt(16))
print("Pi value:", math.pi)

# Method 2: Import specific function
from math import sqrt, pi
print("\nUsing imported functions:")
print("Square root of 25:", sqrt(25))
print("Pi value:", pi)

# Method 3: Import with alias
import random as rnd
print("\nRandom number:", rnd.randint(1, 100))

# Method 4: Import all (not recommended)
# from math import *

# Common modules:
# os - Operating system operations
# sys - System-specific parameters
# random - Generate random numbers
# datetime - Date and time
# json - Work with JSON data
```
