```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Load the CSV file using Pandas.
df = pd.read_csv("Ecommerce_Sales_500_Records.csv")
df
```

```python
# 2. Display first 10 records.
df.head(10)
```

```python
# 3. Check missing values and data types.
df.info()
df.isnull().sum()
df.dtypes
```

```python
# 4. Create Revenue column (Quantity × Price).
df["Revenue"] = df["Quantity"] * df["Price"]
df
```

```python
# 5. Calculate total revenue.
total_revenue = df["Revenue"].sum()
print(total_revenue)
```

```python
# 6. Find top 5 products by revenue.
product_revenue = df.groupby("Product")["Revenue"].sum()
product_revenue
```

```python
# 7. Find category-wise revenue.
category_revenue = df.groupby("Category")["Revenue"].sum()
category_revenue
```

```python
# 8. Find monthly sales trend.

monthly_revenue = df.groupby("Month")["Revenue"].sum()
monthly_revenue
```

```python
# 9. Calculate mean, median, and standard deviation using NumPy.
print(np.mean(df["Quantity"]))
print(np.median(df["Quantity"]))
print(np.std(df["Quantity"]))
```

```python
# 10. Create line chart, bar chart, pie chart, and histogram using Matplotlib.
monthly_revenue.plot(kind="line")
```

```python
product_revenue.plot(kind="bar")
```

```python
category_revenue.plot(kind="pie", autopct="%1.1f%%")
```
