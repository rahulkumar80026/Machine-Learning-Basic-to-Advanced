```python
# from a pandas dataframe
batsman = pd.read_csv('sharma-kohli.csv')

batsman

plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# labels and title
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Run Score')
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'], batsman['V Kohli'],color='green')
plt.plot(batsman['index'], batsman['RG Sharma'],color="#617f0a")

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Run Score')
```

```python
# line style (dashdot,dashed,dotted) linewidth=3 or 2 or 1.....

plt.plot(batsman['index'], batsman['V Kohli'],color='green',linestyle = 'dashed',linewidth=5)
plt.plot(batsman['index'], batsman['RG Sharma'],color='red',linestyle = 'dotted')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Run Score')
```

```python
# Marker diamond(D,o,) markersize(5,20,10,3)
plt.plot(batsman['index'], batsman['V Kohli'],color='green',linestyle = 'dashed',linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'],color='red',linestyle = 'dotted',marker='D',markersize=5)

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Run Score')
```

```python
# legend --> location

plt.plot(batsman['index'], batsman['V Kohli'],color='green',linestyle = 'dashed',linewidth=3,label='Virat')
plt.plot(batsman['index'], batsman['RG Sharma'],color='red',linestyle = 'dotted',marker='D',label="Rohit Sharma")

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Run Score')

plt.legend(loc='best')  # upper right,upper left, lower right,lower left, center left,right,centre right,lowercentre centre 
```

```python
# limiting  axes

# price = [48000, 50000, 52000, 50000, 58000, 55000,600200]
# year = [2015, 2016, 2017, 2018, 2019, 2020,2021]

# plt.plot(year, price)
# ----------------------------------------------------------------------
price = [48000, 50000, 52000, 50000, 58000, 550000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price)

plt.ylim(0,75000)
plt.xlim(2017,2019)
```

```python
#  Grid
plt.plot(batsman['index'], batsman['V Kohli'],color='green',linestyle = 'dashed',linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'],color='red',linestyle = 'dotted',marker='D')

plt.title('Rohit Sharma vs Virat Kohli Carrer Comparison')
plt.xlabel('Season')
plt.ylabel('Run Score')

plt.grid()
```

Bar Chart

> Bivariate Analysis

> Numerical vs Categorical

> Use Case - Aggregate analysis of groups

```python
# simple bar chart
children =[10,20,40,10,50]
colors = ['red','Blue','green','yellow','pink']

plt.bar(colors,children,color='pink')
```

```python
# horizontal bar using # colors and label

plt.barh(colors,children,color='pink')
```

```python
# colors and label
df =pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
# multiple bar chart & xticks
np.arange(df.shape[0])
# plt.bar(np.arange(df.shape[0]) -0.5,df['2015'])
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]) ,df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()

#change color
# plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color = 'red')
```

```python
# a problem
children =[10,20,40,10,50]
colors = ['red Blue green yellow pink','red Blue black yellow pink','red Blue green yellow green','red yellow green yellow pink','red Blue black yellow pink']

plt.bar(colors,children,color='pink')
plt.xticks(rotation='vertical')
plt.show()
```

```python
# Stack bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016']+ df['2017']),label='2015')

plt.legend()
plt.show()
```

Pie chart

> Univariate/Bivariate Analysis 

> Categorical vs numerical

> Use case To find contribution on a standard scale

```python
# simple data 
data = [23,45,100,20,40]
subject = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subject)
plt.show()
```

```python
#  dataset
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'])
plt.show()
```

```python
# percentage and colors
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])

plt.show()
```

```python
# explode shadow 
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0,0,0,0,0.1],shadow=True)

plt.show()
```
